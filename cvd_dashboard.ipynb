```python
import pandas as pd
df = pd.read_csv("Heart Disease Health Indicators Dataset.csv").drop_duplicates()

# Engineered features from your paper
df["Health_Risk_Score"] = df.HighBP + df.HighChol + (df.Diabetes > 0).astype(int)
df["BMI_BP_Interaction"] = df.BMI * df.HighBP
df["BMI_Category"] = pd.cut(df.BMI, [0,18.5,25,30,100],
                            labels=["Underweight","Normal","Overweight","Obese"])

# Readable labels
age_map = dict(zip(range(1,14), ["18-24","25-29","30-34","35-39","40-44","45-49",
          "50-54","55-59","60-64","65-69","70-74","75-79","80+"]))
df["Age_Group"] = df.Age.map(age_map)
df["Sex_Label"] = df.Sex.map({0:"Female",1:"Male"})
df["GenHlth_Label"] = df.GenHlth.map({1:"Excellent",2:"Very good",3:"Good",4:"Fair",5:"Poor"})

# Surrogate-tree pathways (simplified, see note below)
def pathway(r):
    if r.Stroke == 1: return "Override (Stroke)"
    if r.Age >= 10 and r.Sex == 1 and r.BMI_BP_Interaction >= 30: return "High-risk"
    if r.Age <= 9 and r.BMI_BP_Interaction < 30: return "Low-risk"
    return "Intermediate"
df["Risk_Pathway"] = df.apply(pathway, axis=1)

df.to_csv("cvd_tableau.csv", index=False)
```
